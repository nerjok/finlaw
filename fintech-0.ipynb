```python
def busima_verte(suma, norma, metai):
    return suma * (1 + norma) ** metai

try:
    suma = float(input("Pradinė suma: "))
    norma = float(input("Metinė norma (%): ")) / 100
    metai = int(input("Metai: "))

    if suma < 0 or metai < 0:
        raise ValueError("Suma ir metai negali būti neigiami.")

    rezultatas = busima_verte(suma, norma, metai)
    print(f"Po {metai} m.: {rezultatas:.2f} €")
except ValueError as klaida:
    print(f"Neteisingi duomenys: {klaida}")
```
